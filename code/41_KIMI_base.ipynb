# 41 KIMI base

Base example of how to query KIMI directly (without using open-webui).

```python
from dotenv import load_dotenv
import os

load_dotenv()  # Automatically finds .env file
kimi_api_key = os.environ["MOONSHOT_API_KEY"]

from openai import OpenAI
```

```python
MODEL = 'kimi-k2.6'
client = OpenAI(
    api_key=kimi_api_key,
    # Use the base URL shown in your platform console.
    base_url="https://api.moonshot.ai/v1",
    timeout=60,
)

response = client.chat.completions.create(
    model=MODEL,  # Exact model ID from your console
    messages=[
        {"role": "user", "content": "Explain DNS in one paragraph."}
    ],
    # temperature=0.2,
)

print(response.choices[0].message.content)
```
