```python
keywords = ['Travel & Tourism', 'Climate Change', 'Cryptocurrency']
all_news = []

for kw in keywords:
    news = gn.get_news(kw)
    for article in news:
        article['keyword'] = kw  # tag with keyword
    all_news.extend(news)

df_keywords = pd.DataFrame(all_news)
df_keywords.head()
```
