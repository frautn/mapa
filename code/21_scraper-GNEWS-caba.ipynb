```python
from googlenewsdecoder import gnews_decoder
from newspaper import Article

url = "https://news.google.com/rss/articles/CBMi2gFBVV95cUxNVzY4VnlabDFEd2Q2TUZ4bVdEZVJFdGdwNktNUERvMFVxSUFpTDI1LUZ6UENQb0pXSk5LQU90VFlxalVrVXluZGJyQmhBVmxvUGR1MEN6YmV1M091aUtFTWFxSnNTRXdyUlk1Rmt3WWhBMlN2N2IyTzdRYjRkTWdsQ0NZVmZpMUVENFh2RWZISUdfdzNTYktlcFNaNGRhM0FrVk82MGU1Y181OFpfckJ0dDVBQXpfS1ltTm9IanE3M09yNW1RY3hGcEQ4a2xxTzBWOEt4eGI4eDRiZ9IB2gFBVV95cUxNVzY4VnlabDFEd2Q2TUZ4bVdEZVJFdGdwNktNUERvMFVxSUFpTDI1LUZ6UENQb0pXSk5LQU90VFlxalVrVXluZGJyQmhBVmxvUGR1MEN6YmV1M091aUtFTWFxSnNTRXdyUlk1Rmt3WWhBMlN2N2IyTzdRYjRkTWdsQ0NZVmZpMUVENFh2RWZISUdfdzNTYktlcFNaNGRhM0FrVk82MGU1Y181OFpfckJ0dDVBQXpfS1ltTm9IanE3M09yNW1RY3hGcEQ4a2xxTzBWOEt4eGI4eDRiZw?oc=5"


try:
    # 1. Decode the Google link into the actual publisher URL
    decoded = gnews_decoder(url)
    real_url = decoded.get("decoded_url")
    print(f"Decoded Publisher URL: {real_url}")

    # 2. Extract full text using newspaper3k
    article = Article(real_url)
    article.download()
    article.parse()

    print("\n--- Article Details ---")
    print(f"Title: {article.title}")
    print(f"Publish Date: {article.publish_date}")
    print("\nText snippet:")
    print(article.text[:500] + "...\n")

except Exception as e:
    print(f"Error fetching article: {e}")
```
